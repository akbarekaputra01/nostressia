```python
import pandas as pd

# ==== CONFIG ====
INPUT_PATH = "stress_forecast.csv"

# mapping user_id -> nama file
USER_FILES = {
    1: "Our Data - Akbar.csv",
    2: "Our Data - Kaleb.csv",
    3: "Our Data - Epin.csv",
    4: "Our Data - Adel.csv",
    5: "Our Data - Sye.csv",
}

# ==== LOAD ====
df = pd.read_csv(INPUT_PATH)

# Pastikan kolom date bisa diparse
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Kalau ada date yang gagal parse, buang agar tidak bikin error output
df = df.dropna(subset=["date"]).copy()

# ==== TRANSFORM (schema baru) ====
out = pd.DataFrame({
    "Tahun": df["date"].dt.year.astype(int),
    "Bulan": df["date"].dt.strftime("%B"),         # contoh: November
    "Tanggal": df["date"].dt.day.astype(int),

    "Study Hours Per Day (Normalnya Weekday 6)": df["study_hour_per_day"],
    "Extracurricular Hours Per Day": df["extracurricular_hour_per_day"],
    "Sleep Hours Per Day": df["sleep_hour_per_day"],
    'Social Hours Per Day (Normalnya 2, kerkom/belajar bareng dihitung juga)': df["social_hour_per_day"],
    "Physical Activity Hours Per Day (Normalnya 0.5)": df["physical_activity_hour_per_day"],

    "GPA": df["gpa"],
    "Stress Level (1-3)": df["stress_level"],
})

# Simpan juga user_id untuk splitting (tidak ikut ke file final)
out["user_id"] = df["user_id"]

# ==== SPLIT & SAVE ====
for uid, filename in USER_FILES.items():
    user_df = out[out["user_id"] == uid].drop(columns=["user_id"]).copy()

    # Optional: rapihin tipe data angka (biar 0.0 jadi 0 kalau memang integer)
    # Tapi kalau kamu mau tetap float, hapus bagian ini.
    # user_df = user_df.apply(pd.to_numeric, errors="ignore")

    user_df.to_csv(filename, index=False, encoding="utf-8-sig")
    print(f"Saved: {filename} ({len(user_df)} rows)")
```
